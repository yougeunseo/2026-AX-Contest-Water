```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# ⚙️ 한글 폰트 깨짐 방지 설정
# ---------------------------------------------------------
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False 

print("🤖 머신러닝 AI 모델 학습을 시작합니다...\n")

# 1. 학습용 데이터 불러오기 (👑 드디어 찾은 완벽한 열쇠: 탭 분리 + cp949)
df = pd.read_csv('./data/대청호_ML_학습용데이터.csv', sep='\t', encoding='cp949')

# 2. 문제집(X)과 정답지(y) 쪼개기
# '조사일'은 날짜 형태라 AI가 계산할 수 없어서 제외하고, '남조류'는 정답지니까 제외!
X = df.drop(columns=['조사일', '유해남조류 세포수 (cells/㎖)'])
y = df['유해남조류 세포수 (cells/㎖)']

# 3. 공부용(Train)과 시험용(Test) 데이터 나누기
# 시간 흐름이 중요하니까 섞지 않고(shuffle=False) 과거 80%로 공부, 최신 20%로 시험!
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print(f"📚 AI 공부용 데이터: {len(X_train)}세트")
print(f"📝 AI 시험용 데이터: {len(X_test)}세트\n")

# 4. 랜덤 포레스트 AI 모델 소환 및 학습
model = RandomForestRegressor(n_estimators=100, random_state=42)

print("🧠 AI가 열심히 기출문제를 풀며 공식을 찾고 있습니다... (잠시만 대기!)")
model.fit(X_train, y_train)

# 5. 숨겨둔 20% 데이터로 모의고사 채점하기
predictions = model.predict(X_test)
r2 = r2_score(y_test, predictions)

print("\n🎉 학습 완료! 모의고사 채점 결과:")
print(f"👉 AI의 예측 설명력(R² Score): {r2 * 100:.2f}%")

# 6. 실제 정답과 AI의 예측값 비교 그래프 그리기
plt.figure(figsize=(12, 5))
plt.plot(y_test.values, label='실제 측정된 남조류 수치 (Real)', color='blue', alpha=0.6)
plt.plot(predictions, label='AI가 예측한 수치 (Predict)', color='red', alpha=0.8, linestyle='--')
plt.title('대청호 남조류 세포수: 실제값 vs AI 예측값 비교', fontsize=15, fontweight='bold')
plt.xlabel('시간 흐름 (최신 20% 데이터 구간)', fontsize=12)
plt.ylabel('유해남조류 세포수 (cells/㎖)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# ⚙️ 한글 폰트 깨짐 방지 설정
# ---------------------------------------------------------
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False 

print("🔍 데이터 로드 및 AI 모델 재학습 중...")

# 1. 데이터 로드 및 X, y, 모델 다시 세팅!
df = pd.read_csv('./data/대청호_ML_학습용데이터.csv', sep='\t', encoding='cp949')
X = df.drop(columns=['조사일', '유해남조류 세포수 (cells/㎖)'])
y = df['유해남조류 세포수 (cells/㎖)']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# 2. 랜덤 포레스트 AI 초고속 훈련
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

print("🧠 AI의 머릿속 분석 완료! 중요도 그래프를 출력합니다.\n")

# 3. 특성 중요도(Feature Importance) 뽑아내기
importances = rf_model.feature_importances_
feature_names = X.columns
fi_df = pd.DataFrame({'변수명': feature_names, '중요도': importances})
fi_df = fi_df.sort_values(by='중요도', ascending=True)

# 4. 막대그래프 그리기
plt.figure(figsize=(10, 8))
plt.barh(fi_df['변수명'], fi_df['중요도'], color='teal', alpha=0.8)

plt.title('🧠 대청호 녹조 예측 핵심 영향 변수 (Feature Importance)', fontsize=16, fontweight='bold')
plt.xlabel('중요도 (AI 예측 기여도, 합산 1.0)', fontsize=12)
plt.ylabel('입력된 환경 변수', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)

# 숫자 표기 (막대 끝에 소수점 3자리까지 표시)
for index, value in enumerate(fi_df['중요도']):
    plt.text(value + 0.005, index, f'{value:.3f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# ⚙️ 한글 폰트 깨짐 방지 설정
# ---------------------------------------------------------
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False 

print("🚨 [긴급 수정] AI의 컨닝 페이퍼를 압수하고 정정당당하게 재학습합니다...\n")

# 1. 데이터 로드
df = pd.read_csv('./data/대청호_ML_학습용데이터.csv', sep='\t', encoding='cp949')

# 2. ❌ 컨닝 페이퍼(데이터 누수) 열 찾아서 싹 다 버리기!
drop_cols = ['조사일', '유해남조류 세포수 (cells/㎖)']
for col in df.columns:
    # 우점종, 조류독소, 냄새물질이라는 단어가 들어간 열은 싹 다 정답지(X)에서 제외!
    if '우점종' in col or '조류독소' in col or '냄새물질' in col:
        drop_cols.append(col)

X = df.drop(columns=drop_cols)
y = df['유해남조류 세포수 (cells/㎖)']

# 3. 데이터 분할 및 모델 훈련 (정정당당한 모의고사)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# 4. 진짜 예측 점수 확인
real_r2 = r2_score(y_test, rf_model.predict(X_test))
print(f"👉 컨닝 페이퍼 압수 후 진짜 예측 설명력(R² Score): {real_r2 * 100:.2f}%\n")

# 5. 특성 중요도 다시 뽑기
importances = rf_model.feature_importances_
feature_names = X.columns
fi_df = pd.DataFrame({'변수명': feature_names, '중요도': importances})
fi_df = fi_df.sort_values(by='중요도', ascending=True)

# 6. 막대그래프 그리기
plt.figure(figsize=(10, 8))
plt.barh(fi_df['변수명'], fi_df['중요도'], color='coral', alpha=0.8) # 색상도 찐 최종 느낌으로 변경!

plt.title('🧠 대청호 녹조 예측 진짜 핵심 영향 변수 (Feature Importance)', fontsize=16, fontweight='bold')
plt.xlabel('중요도 (AI 예측 기여도, 합산 1.0)', fontsize=12)
plt.ylabel('입력된 환경 변수 (컨닝 데이터 제거 완료)', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)

for index, value in enumerate(fi_df['중요도']):
    plt.text(value + 0.005, index, f'{value:.3f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

# ---------------------------------------------------------
# ⚙️ 한글 폰트 깨짐 방지 설정
# ---------------------------------------------------------
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False 

print("✨ 근서의 아이디어 반영 중: 불필요한 변수 제거 및 이름 변경...\n")

# 1. 데이터 로드 및 컨닝 페이퍼(정답 데이터) 제거
df = pd.read_csv('./data/대청호_ML_학습용데이터.csv', sep='\t', encoding='cp949')
drop_cols = ['조사일', '유해남조류 세포수 (cells/㎖)']
for col in df.columns:
    if '우점종' in col or '조류독소' in col or '냄새물질' in col:
        drop_cols.append(col)
X = df.drop(columns=drop_cols)

# =========================================================
# 💡 [핵심 튜닝] 근서 아이디어 적용 구간!
# =========================================================
# ① 중요도 1% 미만인 '채수위치' 쓰레기 변수 3형제 과감하게 삭제!
X = X.drop(columns=['채수위치_문의', '채수위치_추동', '채수위치_회남'])

# ② 두루뭉술한 '과거'를 전문적인 '7일 전(직전)'으로 이름 변경!
# (수질 측정망 데이터가 보통 주간 단위이므로 7일 전으로 표기)
X = X.rename(columns={
    '과거_남조류(cells/㎖)': '7일 전_남조류(cells/㎖)',
    '과거_수온(℃)': '7일 전_수온(℃)'
})
# =========================================================

y = df['유해남조류 세포수 (cells/㎖)']

# 2. 모델 재학습 (변수를 줄였으니 더 빠르고 깔끔하게 돌아감!)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# 3. 특성 중요도 추출
importances = rf_model.feature_importances_
feature_names = X.columns
fi_df = pd.DataFrame({'변수명': feature_names, '중요도': importances})
fi_df = fi_df.sort_values(by='중요도', ascending=True)

# 4. 환골탈태한 중요도 막대그래프 그리기
plt.figure(figsize=(10, 7)) # 불필요한 변수가 줄어서 세로 길이를 살짝 줄임
plt.barh(fi_df['변수명'], fi_df['중요도'], color='royalblue', alpha=0.8) # 색상도 전문적인 블루로!

plt.title('대청호 녹조 예측: 진짜 핵심 영향 변수 (Feature Selection 완료)', fontsize=16, fontweight='bold')
plt.xlabel('중요도 (AI 예측 기여도, 합산 1.0)', fontsize=12)
plt.ylabel('최적화된 환경 변수 입력값', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)

for index, value in enumerate(fi_df['중요도']):
    plt.text(value + 0.005, index, f'{value:.3f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()
```
