```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# ⚙️ 한글 폰트 깨짐 방지 설정
# ---------------------------------------------------------
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False 

print("🤖 머신러닝 AI 모델 학습을 시작합니다...\n")

# 1. 학습용 데이터 불러오기 (👑 드디어 찾은 완벽한 열쇠: 탭 분리 + cp949)
df = pd.read_csv('./data/대청호_ML_학습용데이터.csv', sep='\t', encoding='cp949')

# 2. 문제집(X)과 정답지(y) 쪼개기
# '조사일'은 날짜 형태라 AI가 계산할 수 없어서 제외하고, '남조류'는 정답지니까 제외!
X = df.drop(columns=['조사일', '유해남조류 세포수 (cells/㎖)'])
y = df['유해남조류 세포수 (cells/㎖)']

# 3. 공부용(Train)과 시험용(Test) 데이터 나누기
# 시간 흐름이 중요하니까 섞지 않고(shuffle=False) 과거 80%로 공부, 최신 20%로 시험!
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

print(f"📚 AI 공부용 데이터: {len(X_train)}세트")
print(f"📝 AI 시험용 데이터: {len(X_test)}세트\n")

# 4. 랜덤 포레스트 AI 모델 소환 및 학습
model = RandomForestRegressor(n_estimators=100, random_state=42)

print("🧠 AI가 열심히 기출문제를 풀며 공식을 찾고 있습니다... (잠시만 대기!)")
model.fit(X_train, y_train)

# 5. 숨겨둔 20% 데이터로 모의고사 채점하기
predictions = model.predict(X_test)
r2 = r2_score(y_test, predictions)

print("\n🎉 학습 완료! 모의고사 채점 결과:")
print(f"👉 AI의 예측 설명력(R² Score): {r2 * 100:.2f}%")

# 6. 실제 정답과 AI의 예측값 비교 그래프 그리기
plt.figure(figsize=(12, 5))
plt.plot(y_test.values, label='실제 측정된 남조류 수치 (Real)', color='blue', alpha=0.6)
plt.plot(predictions, label='AI가 예측한 수치 (Predict)', color='red', alpha=0.8, linestyle='--')
plt.title('대청호 남조류 세포수: 실제값 vs AI 예측값 비교', fontsize=15, fontweight='bold')
plt.xlabel('시간 흐름 (최신 20% 데이터 구간)', fontsize=12)
plt.ylabel('유해남조류 세포수 (cells/㎖)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
from xgboost import XGBRegressor
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

# ---------------------------------------------------------
# ⚙️ 한글 폰트 깨짐 방지 설정
# ---------------------------------------------------------
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False 

print("🚀 튀는 값 잡는 독종 인공지능, XGBoost 출격!\n")

# 1. XGBoost 모델 소환 
# (트리 200개, 학습률 0.1로 고점 돌파를 위한 세팅!)
xgb_model = XGBRegressor(n_estimators=200, learning_rate=0.1, random_state=42)

print("🧠 저번 모의고사에서 틀린 '고점(Peak)' 집중 학습 중... (잠시만 대기!)")
xgb_model.fit(X_train, y_train)

# 2. 시험용 데이터로 모의고사 채점하기
xgb_predictions = xgb_model.predict(X_test)
xgb_r2 = r2_score(y_test, xgb_predictions)

print("\n🎉 학습 완료! XGBoost 채점 결과:")
print(f"👉 XGBoost 예측 설명력(R² Score): {xgb_r2 * 100:.2f}%")

# 3. 실제 정답과 XGBoost의 예측값 비교 그래프 그리기
plt.figure(figsize=(12, 5))
plt.plot(y_test.values, label='실제 측정된 남조류 수치 (Real)', color='blue', alpha=0.5, linewidth=2)
plt.plot(xgb_predictions, label='XGBoost 예측한 수치 (Predict)', color='red', alpha=0.9, linestyle='--')

plt.title('대청호 남조류 세포수: XGBoost 고점 돌파 예측 결과', fontsize=15, fontweight='bold')
plt.xlabel('시간 흐름 (최신 20% 데이터 구간)', fontsize=12)
plt.ylabel('유해남조류 세포수 (cells/㎖)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
