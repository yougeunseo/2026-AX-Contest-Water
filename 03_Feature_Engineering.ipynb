```python
import pandas as pd

print("🚀 AI를 위한 특급 힌트 데이터 제작 중...")

# 1. 시각화 때 썼던 마스터 데이터 불러오기
df = pd.read_csv('./data/대청호 조류 전처리 완료 통합 데이터.csv')
df['조사일'] = pd.to_datetime(df['조사일'])

# 2. 정렬 (시간 순서대로 과거 데이터를 가져와야 하니까 필수!)
df = df.sort_values(by=['채수위치', '조사일']).reset_index(drop=True)

# ---------------------------------------------------------
# ⭐ [핵심 1] 파생 변수(Lag Feature) 만들기
# "지점별로 '저번 주(직전 측정일)'의 수온과 남조류 수치를 끌어와서 힌트로 주자!"
# ---------------------------------------------------------
# .shift(1)은 엑셀에서 한 칸씩 밑으로 밀어 내리는 거랑 똑같아!
df['과거_수온(℃)'] = df.groupby('채수위치')['수온(℃)'].shift(1)
df['과거_남조류(cells/㎖)'] = df.groupby('채수위치')['유해남조류 세포수 (cells/㎖)'].shift(1)

# 첫 번째 날짜는 '과거' 데이터가 없으니까 빈칸(NaN)이 생겨버려. 이건 깔끔하게 삭제!
df = df.dropna()

# ---------------------------------------------------------
# ⭐ [핵심 2] 문자를 숫자로 변환 (One-Hot Encoding)
# "AI는 '문의', '추동' 같은 한글을 못 읽어. 0과 1로 번역해주자!"
# ---------------------------------------------------------
# 채수위치(문의, 추동, 회남)를 각각의 컬럼으로 쪼개서 해당하면 1, 아니면 0으로 만들어줌
df = pd.get_dummies(df, columns=['채수위치'], drop_first=False)

# 이제 진짜 필요 없는 문자 데이터(분류, 지점명)는 버리기
df = df.drop(columns=['분류', '지점명'])

# 3. AI가 먹기 좋게 완성된 최종 학습용 데이터 저장!
save_path = './data/대청호_ML_학습용데이터.csv'
df.to_csv(save_path, index=False, encoding='utf-8-sig')

print(f"🎉 완벽해! AI 학습용 최종 데이터가 저장됐어: {save_path}")
display(df.head())
```
